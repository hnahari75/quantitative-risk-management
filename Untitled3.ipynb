```python
import yfinance as yf

# 1. Create a list of our US Proxy tickers from the map we built
proxies = list(universe_map.keys())

# 2. Download the data
# auto_adjust=True: This is the 'modern' way. It automatically
# handles dividends and stock splits so the 'Close' price is 100% accurate.
print(f"Fetching data for {len(proxies)} assets...")
raw_data = yf.download(
    tickers=proxies,
    start="2007-01-01",
    auto_adjust=True,
    progress=False
)

# 3. Professional Data Summary
print("\n--- BLOCK 1.2: DATA DOWNLOAD COMPLETE ---")
print(f"Total Rows (Days): {len(raw_data)}")
print(f"Total Columns: {raw_data.shape[1]}")

# Display the first few rows of just the 'Close' prices
# This confirms the data is 'clean' and present.
display(raw_data['Close'].head().style.format("{:.2f}"))
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

# 1. Extract only the 'Close' prices (the most important slice for quants)
prices = raw_data['Close']

# 2. Visualize Data Completeness
# This heatmap is a professional way to see exactly when each asset 'started'
plt.figure(figsize=(12, 5))
sns.heatmap(prices.isnull(), yticklabels=False, cbar=False, cmap='rocket')
plt.title("Block 1.3: Data Completeness Audit (Dark = Data | Light = Missing)", fontweight='bold')
plt.show()

# 3. Modern Cleaning: The 'Structural' Filter
# We keep only the rows where at least 80% of our assets have data.
# This prevents our AI from training on 'empty air' before most ETFs existed.
threshold = int(len(prices.columns) * 0.8)
prices_filtered = prices.dropna(thresh=threshold)

# 4. Handle remaining minor gaps (holidays, exchange glitches)
# 'ffill' means forward-fill: carry the last known price forward.
prices_cleaned = prices_filtered.ffill()

print(f"Audit Complete.")
print(f"Original Start: {prices.index[0].date()} | Cleaned Start: {prices_cleaned.index[0].date()}")
print(f"Total gaps remaining: {prices_cleaned.isnull().sum().sum()}")
```

```python
# 1. The 'Hard Drop': This removes any row that still has a single missing value.
# This ensures our 'Matrix' is 100% solid math with no 'NaN' ghosts.
prices_final = prices_cleaned.dropna()

# 2. Professional Audit Output
print("--- BLOCK 1.3b: FINAL SURGICAL CLEAN ---")
print(f"New 'Solid' Start Date: {prices_final.index[0].date()}")
print(f"Total days of perfect data: {len(prices_final)}")
print(f"Gaps remaining: {prices_final.isnull().sum().sum()}")

# 3. Visual Confirmation: The 'Price Highway'
# Let's plot 3 very different assets to ensure they are all perfectly aligned.
plt.figure(figsize=(12, 5))
plt.plot(prices_final['SPY'], label='S&P 500 (US Equity)', color='#1f77b4')
plt.plot(prices_final['GLD'], label='Gold (Hard Asset)', color='#ff7f0e')
plt.plot(prices_final['MTUM'], label='Momentum (Youngest Asset)', color='#2ca02c')

plt.title("Visual Audit: Aligned & Cleaned Price History", fontweight='bold')
plt.legend()
plt.grid(True, alpha=0.2)
plt.show()
```

```python
import numpy as np

# 1. Transform Prices to Log Returns
# .shift(1) looks at yesterday's price.
# np.log() calculates the Natural Logarithm.
returns = np.log(prices_final / prices_final.shift(1)).dropna()

# 2. Professional Statistical Audit
print("--- BLOCK 1.4: LOG-RETURN AUDIT ---")
# Displaying the average move and the 'Risk' (Standard Deviation)
stats_summary = pd.DataFrame({
    'Avg Daily Return (%)': returns.mean() * 100,
    'Daily Volatility (%)': returns.std() * 100
})
display(stats_summary.head().style.format("{:.4f}"))

# 3. Visualizing the 'Bell Curve'
# We want to see how often 'Extreme' events happen vs 'Normal' days.
plt.figure(figsize=(10, 5))
plt.hist(returns['SPY'], bins=100, alpha=0.6, color='#1f77b4', label='S&P 500 Returns')
plt.hist(returns['GLD'], bins=100, alpha=0.6, color='#ff7f0e', label='Gold Returns')

plt.title("Visual Audit: The Distribution of Daily Returns", fontweight='bold')
plt.xlabel("Log Return (Force of Move)")
plt.ylabel("Frequency (Number of Days)")
plt.legend()
plt.grid(axis='y', alpha=0.2)
plt.show()
```

```python
# 1. Define the 'Sigma' limit.
# 6 Sigma is the institutional standard for 'extreme but possible'.
# Anything beyond this is usually a data error or a once-in-a-century event.
n_sigma = 6

# 2. Calculate the boundaries for every asset
# This is 'Modern' because it adapts to each ETF's specific personality (volatility)
mean = returns.mean()
std = returns.std()

lower_bound = mean - (n_sigma * std)
upper_bound = mean + (n_sigma * std)

# 3. Apply Winsorization (The Scrub)
# .clip() effectively 'squashes' the outliers into the boundary
returns_scrubbed = returns.clip(lower=lower_bound, upper=upper_bound, axis=1)

# --- PROFESSIONAL AUDIT ---
print("--- BLOCK 1.5: OUTLIER SCRUBBING COMPLETE ---")
# Check if any values were actually changed
outliers_detected = (returns != returns_scrubbed).sum().sum()
print(f"Total 'Extreme Outliers' scrubbed: {outliers_detected}")

# Visualize the 'Scrubbed' tail of the S&P 500
plt.figure(figsize=(10, 4))
plt.plot(returns['SPY'], label='Original', alpha=0.3, color='red')
plt.plot(returns_scrubbed['SPY'], label='Scrubbed (Winsorized)', alpha=0.8, color='#1f77b4')
plt.title("Audit: S&P 500 Returns Before vs After 6-Sigma Scrubbing", fontweight='bold')
plt.legend()
plt.show()
```

```python
# 1. Isolate our Anchor
# HMM requires the data to be in a specific 'column' shape, so we use .values.reshape(-1, 1)
anchor_name = 'SPY'
anchor_returns = returns_scrubbed[anchor_name].values.reshape(-1, 1)

# 2. Visualizing the 'Why': Volatility Clustering
# Modern finance is based on the idea that 'Vol' comes in waves.
# Look at the chart below: when it gets 'messy', that is a Regime Change.
plt.figure(figsize=(12, 5))
plt.plot(returns_scrubbed.index, returns_scrubbed[anchor_name], color='#1f77b4', alpha=0.5)
plt.title(f"Block 2.1: {anchor_name} Returns - Spotting the Clusters", fontweight='bold')
plt.axhline(0, color='black', lw=1)
plt.ylabel("Daily Log Return")
plt.grid(True, alpha=0.2)
plt.show()

print(f"Anchor '{anchor_name}' prepared for Machine Learning.")
```

```python
from hmmlearn.hmm import GaussianHMM

# We define the 'Architecture' of our Weather Station:
# n_components=2: Tell the AI to look for exactly 2 states (Bull vs. Bear)
# covariance_type="full": Allow the AI to be very precise about risk levels
# n_iter=1000: Give the AI enough 'thinking time' to find the best fit
hmm_model = GaussianHMM(
    n_components=2,
    covariance_type="full",
    n_iter=1000,
    random_state=42
)

print("HMM Brain Initialized and waiting for data.")
```

```python
# 1. Fit the model to the anchor returns (SPY)
# We use .reshape(-1, 1) because the HMM expects a 2D array
hmm_model.fit(anchor_returns)

# 2. Predict the 'Hidden States' (0 or 1) for every day in our history
hidden_states = hmm_model.predict(anchor_returns)

print("Training complete. The AI has categorized every day into one of two 'Regimes'.")
```

```python
# 1. Calculate the Expected Tail Loss (CVaR) for both 'Hidden States'
# This ensures we label the 'Bear' based on actual downside danger
state_cvars = []
for i in range(2):
    state_data = anchor_returns[hidden_states == i]
    # We use the 95% CVaR (worst 5% of days)
    var_95 = np.percentile(state_data, 5)
    cvar_95 = state_data[state_data <= var_95].mean()
    state_cvars.append(cvar_95)

# 2. Identify the Bear Regime (the state with the most negative/worst CVaR)
bear_regime_index = np.argmin(state_cvars)

print(f"Regime Detection Results (CVaR-Based):")
print(f" - State {1 - bear_regime_index} (Bull) Expected Tail Loss: {state_cvars[1-bear_regime_index]*100:.2f}%")
print(f" - State {bear_regime_index} (Bear) Expected Tail Loss: {state_cvars[bear_regime_index]*100:.2f}%")
```

```python
# 1. Create the plot
plt.figure(figsize=(12, 6))

# 2. Loop through each state and plot it with a specific color
for i in range(2):
    mask = (hidden_states == i)
    color = "red" if i == bear_regime_index else "green"
    label = "Bear State" if i == bear_regime_index else "Bull State"

    plt.scatter(returns_scrubbed.index[mask],
                returns_scrubbed[anchor_name].iloc[mask],
                color=color, label=label, s=4, alpha=0.6)

plt.title(f"Block 2.2e: {anchor_name} Regime Mapping", fontweight='bold')
plt.legend()
plt.grid(True, alpha=0.2)
plt.show()
```

```python
# 1. Project all 10,000 simulated portfolios onto our historical Bull data
# This creates a 'History' for every single candidate (10,000 portfolios)
historical_bull_matrix = regime_returns.values
# Matrix math: (Days x Assets) @ (Assets x 10,000 Portfolios)
port_histories = np.dot(historical_bull_matrix, all_weights.T)

# 2. Calculate the 95% CVaR for each of the 10,000 candidates
cvar_arr = []
for i in range(num_portfolios):
    p_daily_returns = port_histories[:, i]
    # Find the 5% threshold (VaR)
    v_95 = np.percentile(p_daily_returns, 5)
    # Average the returns that fell BELOW that threshold (CVaR)
    c_95 = p_daily_returns[p_daily_returns <= v_95].mean()
    cvar_arr.append(c_95)

cvar_arr = np.array(cvar_arr)

print(f"--- BLOCK 3.3: CVaR STRESS TEST COMPLETE ---")
print(f"Worst-case CVaR found: {cvar_arr.min()*100:.2f}% (Deepest hole)")
print(f"Best-case CVaR found: {cvar_arr.max()*100:.2f}% (Strongest shield)")
```

```python
# 1. Calculate the 'Tail Ratio' for all 10,000 candidates
# This is like a Sharpe Ratio, but specifically for extreme risk
tail_ratios = ret_arr / np.abs(cvar_arr)

# 2. Find the index of the highest ratio
best_idx = np.argmax(tail_ratios)

# 3. Extract the 'Winning' stats
winner_return = ret_arr[best_idx]
winner_cvar = cvar_arr[best_idx]
winner_weights = all_weights[best_idx]

print(f"--- BLOCK 3.4: OPTIMAL WINNER FOUND ---")
print(f"Winning Expected Annual Return: {winner_return*100:.2f}%")
print(f"Winning Expected Tail Loss (CVaR): {winner_cvar*100:.2f}%")
print(f"Efficiency Score (Tail Ratio): {tail_ratios[best_idx]:.2f}")
```

```python
import numpy as np

# 1. Define the Backtest Window (4 Years: 2019-01-01 to 2022-12-31)
start_date = '2019-01-01'
end_date = '2022-12-31'

# Slice the scrubbed returns and hidden states to match our 4-year timeline
mask = (returns_scrubbed.index >= start_date) & (returns_scrubbed.index <= end_date)
bt_returns = returns_scrubbed.loc[mask]
bt_states = hidden_states[mask]

# 2. Define the Algorithm's Target Weights for each Regime
# NOTE: Update these dictionaries with your algorithm's exact allocations.
# Weights must sum to 1.0 (100%)
bull_regime_weights = {
    "VUSA.DE": 0.50,  # 50% S&P 500
    "SXRV.DE": 0.50   # 50% Nasdaq
}

bear_regime_weights = {
    "DTLA.L": 0.50,   # 50% Long-term Treasury Bonds
    "4GLD.DE": 0.50   # 50% Gold
}

# 3. Initialize Trading 212 Simulation Parameters
initial_capital = 200.0
monthly_deposit = 200.0

portfolio_value = initial_capital
total_invested = initial_capital

portfolio_history = []
invested_history = []

# Track the month to trigger monthly deposits
current_month = bt_returns.index[0].month

# 4. Run the Step-by-Step Simulation
for i in range(len(bt_returns)):
    current_date = bt_returns.index[i]
    current_state = bt_states[i]

    # Check for Month-End / Start of new month to apply the 200 EUR deposit
    if current_date.month != current_month:
        portfolio_value += monthly_deposit
        total_invested += monthly_deposit
        current_month = current_date.month

    # Determine which weights to use based on the HMM's regime prediction
    if current_state == bear_regime_index:
        active_weights = bear_regime_weights
    else:
        active_weights = bull_regime_weights

    # Calculate the weighted daily return of the portfolio
    daily_portfolio_return = 0.0
    for ticker, weight in active_weights.items():
        # Ensure the ticker exists in our return data
        if ticker in bt_returns.columns:
            daily_portfolio_return += weight * bt_returns[ticker].iloc[i]

    # Apply the daily return to the portfolio value (converting log return back to simple multiplier)
    portfolio_value = portfolio_value * np.exp(daily_portfolio_return)

    # Record the day's results
    portfolio_history.append(portfolio_value)
    invested_history.append(total_invested)

# 5. Professional Visual Output
plt.figure(figsize=(12, 6))

plt.plot(bt_returns.index, portfolio_history, label='Algorithm Portfolio Value', color='#2ca02c', linewidth=2)
plt.plot(bt_returns.index, invested_history, label='Total Cash Deposited (Zero Return Base)', color='gray', linestyle='--', linewidth=1.5)

# Calculate final metrics
total_profit = portfolio_history[-1] - invested_history[-1]
roi_percentage = (total_profit / invested_history[-1]) * 100

plt.title(f"Block 3: 4-Year Algorithm Backtest (Start: €{initial_capital} | €{monthly_deposit}/mo)", fontweight='bold')
plt.xlabel("Date")
plt.ylabel("Account Value (EUR)")
plt.legend()
plt.grid(True, alpha=0.2)

# Fill the area showing pure profit
plt.fill_between(bt_returns.index, invested_history, portfolio_history, where=(np.array(portfolio_history) > np.array(invested_history)), color='green', alpha=0.1)
plt.fill_between(bt_returns.index, invested_history, portfolio_history, where=(np.array(portfolio_history) < np.array(invested_history)), color='red', alpha=0.1)

plt.show()

print("--- BLOCK 3.1: PERFORMANCE SUMMARY ---")
print(f"Total Cash Invested: €{invested_history[-1]:,.2f}")
print(f"Final Portfolio Value: €{portfolio_history[-1]:,.2f}")
print(f"Total Net Profit: €{total_profit:,.2f}")
print(f"Return on Invested Capital: {roi_percentage:.2f}%")
```
